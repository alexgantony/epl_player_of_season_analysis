On Not Getting Blocked
There are four techniques that matter here, in order of importance:
1. Respect rate limiting — add delays between requests
The single most important thing. Between every request, sleep for a random interval rather than a fixed one. Fixed delays are easier for servers to detect as bots. Random delays mimic human behaviour.
The concept is: time.sleep(random.uniform(3, 7)) between each page request. For 5 pages this adds maybe 25 seconds total — completely acceptable.
2. Set a proper User-Agent header
By default, requests identifies itself as a Python script. That's an immediate bot signal. Set your request headers to identify as a normal browser. Look up what a standard Chrome User-Agent string looks like and pass it in your request headers.
3. Use a session object
Rather than making independent requests, use requests.Session(). This maintains cookies and connection state across requests, which looks significantly more like a real browser session than repeated cold requests.
4. Check robots.txt first
Visit fbref.com/robots.txt manually before writing any code. This tells you which paths Sports Reference explicitly disallows. If your target URLs are disallowed, you have a decision to make about approach — either use their official data export options or reach out for permission. This is both ethical practice and protects your portfolio from scrutiny.

```python
goalkeeping_path = FBREF_WEBPAGES_PATH / "goalkeeping.html"
standard_stats_path = FBREF_WEBPAGES_PATH / "standard_stats.html"
shooting_path = FBREF_WEBPAGES_PATH / "shooting.html"
playingtime_path = FBREF_WEBPAGES_PATH / "playingtime.html"
misc_stats_path = FBREF_WEBPAGES_PATH / "misc_stats.html"
```

```python
def extract_data_from_table(path):
    with open(path, encoding="utf-8") as f:
        soup = BeautifulSoup(f, "html.parser")
    
    comments = soup.find_all(string=lambda text: isinstance(text, Comment))
    
    table = [comment.strip() for comment in comments if "<table" in comment][0]
    soup_table = BeautifulSoup(table)
    html = str(soup_table.table)

    df = pd.read_html(StringIO(html))[0]

    return df
```

```python
goalkeeping_df = extract_data_from_table(goalkeeping_path)
standard_stats_df = extract_data_from_table(standard_stats_path)
shooting_df = extract_data_from_table(shooting_path)
playingtime_df = extract_data_from_table(playingtime_path)
misc_stats_df = extract_data_from_table(misc_stats_path)
```
