### Data Inspection

```python
def data_inspection(df: pd.DataFrame, df_name: str) -> None:
    """Display basic information about a DataFrame."""

    print("=" * 50)
    print(df_name)
    print("=" * 50)

    print(f"Shape: {df.shape}")
    print(f"Number of duplicate values: {df.duplicated().sum()}")

    missing = df.isna().sum()
    missing = missing[missing > 0]

    if missing.empty:
        print("Missing values: None")
    else:
        print("Missing values:")
        print(missing)
```

### Handling Mid-Season Transfers

Some players represented two clubs during the season and appear in the dataset with club names separated by a forward slash (`/`), for example:

- `Chelsea / Everton`
- `Fulham / West Ham`

For this analysis, these players are assigned to the club listed **after the forward slash**, representing the club they belonged to at the end of the season. This ensures that each player is associated with a single club, avoids duplicate club assignments, and aligns team-based metrics with the player's final club.

```python
fbref_player_df[fbref_player_df["Squad"].str.contains("/")]
```

```python
# Players transferred mid-season are assigned to their final club.
# If a player appears as "Club A / Club B", retain "Club B".
fbref_player_df["Squad"] = fbref_player_df["Squad"].str.split(" / ").str[-1]
```

```python
def compare_player_names(
        fbref_df: pd.DataFrame, 
        analyst_df: pd.DataFrame, 
        fbref_col: str = "Player", 
        analyst_col: str = "player") -> tuple:
    """
    Compare player names between the FBref and The Analyst datasets.

    Displays player names that are unique to each dataset, allowing
    mismatches to be identified and resolved before merging.
    """

    fbref_players = set(fbref_df[fbref_col].str.strip())
    analyst_players = set(analyst_df[analyst_col].str.strip())

    only_in_fbref = fbref_players.difference(analyst_players)
    only_in_analyst = analyst_players.difference(fbref_players)

    print("Only in FBref:")
    print(sorted(only_in_fbref))

    print("\nOnly in The Analyst:")
    print(sorted(only_in_analyst))

    return  only_in_fbref, only_in_analyst
```

### Goalkeeper Dataset Column Selection

The merged goalkeeper dataset contained several redundant, derived, ambiguous, and goalkeeper-irrelevant variables. These columns were removed to simplify the dataset and retain only features that may contribute meaningfully to goalkeeper performance evaluation.

#### Redundant Columns

These variables duplicated information available elsewhere in the dataset.

- `minutes` (replaced by `mins_played`)
- `goals_allowed_on_pitch` (replaced by `goals_conceded`)
- `Born` (replaced by `Age`)
- `Pos` (replaced by `squad_position`)
- `player` (duplicate player identifier)
- `contestantClubName` (duplicate club identifier)

#### Goalkeeper-Irrelevant Columns

The following variables describe attacking or outfield-specific actions that are not meaningful for evaluating goalkeepers.

- `shots_total`
- `shots_per90`
- `shots_on_target`
- `shots_on_target_per90`
- `shots_on_target_pct`
- `goals_per_shot`
- `goals_per_shot_on_target`
- `penalty_kicks_made`
- `penalty_kicks_attempted`
- `offsides`

#### Derived Attacking Metrics

These metrics were removed because they are derived from attacking statistics that are not central to goalkeeper evaluation.

- `non_penalty_goals`
- `goals_per90`
- `non_penalty_goals_per90`
- `goals_assists_per90`
- `non_penalty_goals_assists_per90`

#### Ambiguous Variable

The `ogs` variable from The Analyst dataset could not be conclusively identified using the available documentation or data dictionary. Although exploratory validation confirmed that it does not represent a goalkeeper's own goals, its exact definition could not be verified with sufficient confidence.

To preserve the integrity and interpretability of the canonical dataset, this variable was excluded from further analysis.

- `ogs`

#### Notes

- `goals`, `assists`, and `goals_assists` were retained because goalkeepers can occasionally make meaningful attacking contributions (e.g., providing an assist or scoring a goal).
- Columns were removed only when they were **redundant, derived, irrelevant to goalkeeper evaluation, or could not be confidently interpreted**.
- The final goalkeeper dataset retains only well-defined features that are relevant for subsequent feature engineering and POTSI score construction.
