```python
import pandas as pd
import json
```

```python
from pathlib import Path

ROOT = Path().resolve().parent
RAW_DATA = ROOT / "data" / "raw"
INTERIM_DATA = ROOT / "data" / "interim"
```

```python
with open(Path(RAW_DATA / "the_analyst.json"), encoding="utf-8") as f:
    data = json.load(f)


player_data = data["player"]
groups = ["attack", "possession", "carries", "defending", "goalkeeping"]
dfs = {}

for key in player_data.keys():
    if key in groups:
        for category, df in player_data[key].items():
            dfs[f"{key}_{category}"] = pd.DataFrame(df)
```

```python
comm_cols_to_del.pop(comm_cols_to_del.index("player_id"))
```

```python
attack_nonPenalty = dfs["attack_nonPenalty"].drop(columns=comm_cols_to_del)
```

```python
attack_df = pd.merge(dfs["attack_overall"], attack_nonPenalty, on="player_id")
```
